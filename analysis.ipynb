```python
# =========================================
# 🔥 FINAL RESEARCH PIPELINE (ALL-IN-ONE)
# =========================================

import torch, time, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.metrics import precision_recall_fscore_support
from sklearn.manifold import TSNE

from transformers import AutoTokenizer, AutoModel
from torch.utils.data import Dataset, DataLoader

warnings.filterwarnings("ignore")
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# =========================================
# 1. LOAD DATA
# =========================================
df_stage2 = pd.read_csv('/kaggle/input/datasets/ngohoangdang/df-stage-2-out/df_stage2_out.csv')
df = pd.read_csv('/kaggle/input/datasets/organizations/uciml/sms-spam-collection-dataset/spam.csv', encoding='latin-1')

df = df.rename(columns={'v1':'label','v2':'text'})
df['msg_id'] = np.arange(len(df))


df_stage2 = df_stage2.drop_duplicates(subset=['obfuscated'])
df_stage2 = df_stage2[df_stage2['recover_score'] > 0.2]
msg_map = df.set_index('text')['msg_id'].to_dict()
df_stage2['msg_id'] = df_stage2['original'].map(msg_map)

train_ids, test_ids = train_test_split(df.msg_id, test_size=0.2, random_state=42)
train_ids, val_ids  = train_test_split(train_ids, test_size=0.1, random_state=42)

def create_dataset(ids):
    real = df[df.msg_id.isin(ids)].copy()
    real['data_type'] = 'real'

    syn = df_stage2[df_stage2.msg_id.isin(ids)].copy()
    syn['data_type'] = 'synthetic'
    syn = syn.rename(columns={'obfuscated':'text'})
    syn['label'] = 'spam'

    return pd.concat([
        real[['msg_id','text','label','data_type']],
        syn[['msg_id','text','label','data_type']]
    ])

df_train = create_dataset(train_ids)
df_val   = create_dataset(val_ids)
df_test  = create_dataset(test_ids)

# =========================================
# 2. TOKENIZER + VOCAB
# =========================================
tokenizer = AutoTokenizer.from_pretrained("distilbert-base-uncased")

MAX_LEN = 64
VOCAB_SIZE = 20000

def simple_tokenize(text):
    return str(text).lower().split()[:MAX_LEN]

from collections import Counter
counter = Counter()
for t in df_train.text:
    counter.update(simple_tokenize(t))

vocab = {w:i+1 for i,(w,_) in enumerate(counter.most_common(VOCAB_SIZE))}

def encode(text):
    tokens = [vocab.get(w,0) for w in simple_tokenize(text)]
    return tokens + [0]*(MAX_LEN-len(tokens))

CHARS = list("abcdefghijklmnopqrstuvwxyz0123456789")
char2idx = {c:i+1 for i,c in enumerate(CHARS)}

def encode_char(text):
    arr = [char2idx.get(c,0) for c in str(text.lower())[:128]]
    return arr + [0]*(128-len(arr))

# =========================================
# 3. DATASET
# =========================================
class SpamDataset(Dataset):
    def __init__(self, df):
        self.df = df.reset_index(drop=True)
        self.map = {'ham':0,'spam':1}

    def __len__(self): return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]

        bert_enc = tokenizer(
            row.text,
            truncation=True,
            padding='max_length',
            max_length=64,
            return_tensors='pt'
        )

        return {
            'ids': torch.tensor(encode(row.text)),
            'bert_ids': bert_enc['input_ids'].squeeze(),
            'mask': bert_enc['attention_mask'].squeeze(),
            'char_ids': torch.tensor(encode_char(row.text)),
            'label': torch.tensor(self.map[row.label]),
            'msg_id': torch.tensor(row.msg_id),
            'type': row.data_type
        }

# =========================================
# 4. MODELS
# =========================================
class DistilBERT_Model(torch.nn.Module):
    def __init__(self):
        super().__init__()
        self.bert = AutoModel.from_pretrained("distilbert-base-uncased")
        self.fc = torch.nn.Linear(768,2)

    def forward(self, bert_ids, mask):
        out = self.bert(bert_ids, attention_mask=mask)
        feat = out.last_hidden_state[:,0,:]
        return self.fc(feat), feat

class FastText(torch.nn.Module):
    def __init__(self):
        super().__init__()
        self.emb = torch.nn.Embedding(VOCAB_SIZE+1,128)
        self.fc = torch.nn.Linear(128,2)

    def forward(self, ids):
        x = self.emb(ids)
        feat = x.mean(1)
        return self.fc(feat), feat

class TextCNN(torch.nn.Module):
    def __init__(self):
        super().__init__()
        self.emb = torch.nn.Embedding(VOCAB_SIZE+1,128)
        self.convs = torch.nn.ModuleList(
            [torch.nn.Conv1d(128,128,k) for k in [3,4,5]]
        )
        self.fc = torch.nn.Linear(384,2)

    def forward(self, ids):
        x = self.emb(ids).permute(0,2,1)
        xs = [torch.relu(c(x)).max(2)[0] for c in self.convs]
        feat = torch.cat(xs,1)
        return self.fc(feat), feat

class CharCNN(torch.nn.Module):
    def __init__(self):
        super().__init__()
        self.emb = torch.nn.Embedding(len(char2idx)+1,64)
        self.conv = torch.nn.Sequential(
            torch.nn.Conv1d(64,128,3),
            torch.nn.ReLU(),
            torch.nn.Conv1d(128,128,3),
            torch.nn.ReLU()
        )
        self.fc = torch.nn.Linear(128,2)

    def forward(self, char_ids):
        x = self.emb(char_ids).permute(0,2,1)
        x = self.conv(x)
        feat = x.max(2)[0]
        return self.fc(feat), feat

# =========================================
# 5. FORWARD WRAPPER
# =========================================
def forward_model(model, batch):

    if isinstance(model, DistilBERT_Model):
        return model(
            batch['bert_ids'].to(device),
            batch['mask'].to(device)
        )
    elif isinstance(model, CharCNN):
        return model(batch['char_ids'].to(device))
    else:
        return model(batch['ids'].to(device))

# =========================================
# 6. LOSS
# =========================================
def compute_loss(logits, feat, labels, batch, m):

    loss = torch.nn.functional.cross_entropy(logits, labels)

    msg_ids = batch['msg_id'].to(device)
    same = (msg_ids.unsqueeze(0)==msg_ids.unsqueeze(1)).float()
    same.fill_diagonal_(0)

    if m>=2:
        p = torch.softmax(logits,1)
        log_p = torch.log_softmax(logits,1)
        kl = (p.unsqueeze(1)*(log_p.unsqueeze(1)-log_p.unsqueeze(0))).sum(-1)
        loss += 0.5*(same*kl).sum()/(same.sum()+1e-8)

    if m>=3:
        p = torch.softmax(logits,1)
        loss += 0.1*-(p*torch.log(p+1e-8)).sum(1).mean()

    if m==4:
        f = torch.nn.functional.normalize(feat,dim=1)
        sim = torch.matmul(f,f.T)/0.1

        exp_sim = torch.exp(sim)
        pos = exp_sim*same
        neg = exp_sim*(1-same)

        pos_sum = pos.sum(1)
        valid = (pos_sum>0)

        if valid.sum()>0:
            loss += 0.1*(-torch.log(
                pos_sum[valid]/(pos_sum[valid]+neg.sum(1)[valid]+1e-8)
            ).mean())

    return loss

# =========================================
# 7. TRAIN
# =========================================
def train(model, df, m):

    model.to(device)

    if isinstance(model, DistilBERT_Model):
        opt = torch.optim.AdamW(model.parameters(), lr=3e-5)
    else:
        opt = torch.optim.Adam(model.parameters(), lr=1e-3)

    loader = DataLoader(SpamDataset(df), batch_size=64, shuffle=True)

    for epoch in range(3):
        model.train()

        for b in loader:
            opt.zero_grad()

            logits, feat = forward_model(model, b)

            loss = compute_loss(
                logits, feat,
                b['label'].to(device),
                b, m
            )

            loss.backward()

            # 🔥 chống explode
            if isinstance(model, DistilBERT_Model):
                torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)

            opt.step()

    return model

# =========================================
# 8. EVALUATE (FULL METRICS 🔥)
# =========================================
def evaluate(model, df):

    loader = DataLoader(SpamDataset(df), batch_size=256)

    y_true, y_pred, types = [], [], []

    start = time.time()
    total = 0

    with torch.no_grad():
        for b in loader:
            logits,_ = forward_model(model, b)

            pred = torch.argmax(logits,1).cpu().numpy()

            y_pred.extend(pred)
            y_true.extend(b['label'].numpy())
            types.extend(b['type'])

            total += len(pred)

    total_time = time.time() - start

    y_true = np.array(y_true)
    y_pred = np.array(y_pred)
    types  = np.array(types)

    def calc(mask):
        if mask.sum()==0:
            return (0,0,0)
        return precision_recall_fscore_support(
            y_true[mask],
            y_pred[mask],
            average='binary',
            zero_division=0
        )[:3]

    metrics = {
        "all": calc(np.ones(len(y_true),bool)),
        "real": calc(types=='real'),
        "synthetic": calc(types=='synthetic'),
        "latency": (total_time/total)*1000,
        "throughput": total/total_time
    }

    return metrics

# =========================================
# 9. TSNE (3 COLORS)
# =========================================
def plot_tsne(model, df, title):

    loader = DataLoader(SpamDataset(df), batch_size=256)

    feats, labels, types = [], [], []

    with torch.no_grad():
        for b in loader:
            _, f = forward_model(model, b)
            feats.append(f.cpu())
            labels.extend(b['label'].numpy())
            types.extend(b['type'])

    feats = torch.cat(feats).numpy()
    emb = TSNE(n_components=2).fit_transform(feats)

    labels = np.array(labels)
    types  = np.array(types)

    spam_real = (types=='real') & (labels==1)
    ham_real  = (types=='real') & (labels==0)
    synthetic = (types=='synthetic')

    plt.figure(figsize=(6,6))
    plt.scatter(emb[spam_real,0], emb[spam_real,1], c='red', label='spam_real', alpha=0.6)
    plt.scatter(emb[ham_real,0], emb[ham_real,1], c='blue', label='ham_real', alpha=0.6)
    plt.scatter(emb[synthetic,0], emb[synthetic,1], c='green', label='synthetic', alpha=0.6)

    plt.title(title)
    plt.legend()
    plt.show()

# =========================================
# 10. RUN
# =========================================
models = {
    "FastText": FastText,
    "TextCNN": TextCNN,
    "CharCNN": CharCNN,
    "DistilBERT": DistilBERT_Model
}

results = []

for name, cls in models.items():
    for m in [1,2,3,4]:

        print(f"\n=== {name} M{m} ===")

        train_df_use = df_train if m!=1 else df_train[df_train.data_type=='real']

        model = train(cls(), train_df_use, m)

        res = evaluate(model, df_test)

        print(f"""
ALL       -> F1: {res['all'][2]:.4f}
REAL      -> F1: {res['real'][2]:.4f}
SYNTHETIC -> F1: {res['synthetic'][2]:.4f}

Latency:   {res['latency']:.2f} ms/sample
Throughput:{res['throughput']:.2f} samples/sec
""")

        results.append({
            "Model": f"{name}_M{m}",
            "F1_all": res['all'][2],
            "F1_real": res['real'][2],
            "F1_syn": res['synthetic'][2],
            "Gap": res['real'][2] - res['synthetic'][2],
            "Latency": res['latency'],
            "Throughput": res['throughput']
        })

        plot_tsne(model, df_test, f"{name}_M{m}")

# summary table
df_res = pd.DataFrame(results)
print("\n===== FINAL RESULT =====")
print(df_res.sort_values("F1_all", ascending=False))
```

```python
# =========================================================
# 🔥 CHARCNN-M4 : FIND OPTIMAL SYNTHETIC DATA SIZE
# Goal:
# Determine how much synthetic data should be added for M4 training
# =========================================================

import torch
import time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.metrics import precision_recall_fscore_support
from torch.utils.data import Dataset, DataLoader

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# =========================================================
# 1. LOAD DATA
# =========================================================

df_stage2 = pd.read_csv('/kaggle/input/datasets/ngohoangdang/df-stage-2-out/df_stage2_out.csv')
df = pd.read_csv(
    '/kaggle/input/datasets/organizations/uciml/sms-spam-collection-dataset/spam.csv',
    encoding='latin-1'
)

df = df.rename(columns={'v1':'label', 'v2':'text'})
df['msg_id'] = np.arange(len(df))

df_stage2 = df_stage2.drop_duplicates(subset=['obfuscated'])
df_stage2 = df_stage2[df_stage2['recover_score'] > 0.2]

msg_map = df.set_index('text')['msg_id'].to_dict()
df_stage2['msg_id'] = df_stage2['original'].map(msg_map)

train_ids, test_ids = train_test_split(
    df.msg_id,
    test_size=0.2,
    random_state=42
)

# =========================================================
# 2. CHAR VOCAB
# =========================================================

CHARS = list("abcdefghijklmnopqrstuvwxyz0123456789")
char2idx = {c:i+1 for i,c in enumerate(CHARS)}

def encode_char(text, max_len=128):
    arr = [char2idx.get(c, 0) for c in str(text).lower()[:max_len]]
    return arr + [0] * (max_len - len(arr))

# =========================================================
# 3. DATASET BUILDER
# =========================================================

def create_dataset_with_n_synthetic(ids, n_syn):

    # ----- real -----
    real = df[df.msg_id.isin(ids)].copy()
    real['data_type'] = 'real'

    # ----- synthetic -----
    syn = df_stage2[df_stage2.msg_id.isin(ids)].copy()

    # sample exactly n synthetic
    if len(syn) > n_syn:
        syn = syn.sample(n=n_syn, random_state=42)

    syn['data_type'] = 'synthetic'
    syn = syn.rename(columns={'obfuscated': 'text'})
    syn['label'] = 'spam'

    syn = syn[['msg_id', 'text', 'label', 'data_type']]
    real = real[['msg_id', 'text', 'label', 'data_type']]

    final_df = pd.concat([real, syn]).reset_index(drop=True)

    print(f"Real samples: {len(real)}")
    print(f"Synthetic samples used: {len(syn)}")
    print(f"Total train size: {len(final_df)}")

    return final_df


def create_test_dataset(ids):
    real = df[df.msg_id.isin(ids)].copy()
    real['data_type'] = 'real'

    syn = df_stage2[df_stage2.msg_id.isin(ids)].copy()
    syn['data_type'] = 'synthetic'
    syn = syn.rename(columns={'obfuscated': 'text'})
    syn['label'] = 'spam'

    return pd.concat([
        real[['msg_id','text','label','data_type']],
        syn[['msg_id','text','label','data_type']]
    ]).reset_index(drop=True)


df_test = create_test_dataset(test_ids)

# =========================================================
# 4. DATASET CLASS
# =========================================================

class SpamDataset(Dataset):
    def __init__(self, df):
        self.df = df.reset_index(drop=True)
        self.map = {'ham': 0, 'spam': 1}

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]

        return {
            'char_ids': torch.tensor(
                encode_char(row.text),
                dtype=torch.long
            ),
            'label': torch.tensor(
                self.map[row.label],
                dtype=torch.long
            ),
            'msg_id': torch.tensor(row.msg_id),
            'type': row.data_type
        }

# =========================================================
# 5. CHARCNN MODEL
# =========================================================

class CharCNN(torch.nn.Module):
    def __init__(self):
        super().__init__()

        self.emb = torch.nn.Embedding(
            len(char2idx) + 1,
            64
        )

        self.conv = torch.nn.Sequential(
            torch.nn.Conv1d(64, 128, 3),
            torch.nn.ReLU(),

            torch.nn.Conv1d(128, 128, 3),
            torch.nn.ReLU()
        )

        self.fc = torch.nn.Linear(128, 2)

    def forward(self, char_ids):
        x = self.emb(char_ids).permute(0, 2, 1)
        x = self.conv(x)

        feat = x.max(2)[0]
        logits = self.fc(feat)

        return logits, feat

# =========================================================
# 6. M4 LOSS (CONTRASTIVE)
# =========================================================

def compute_m4_loss(logits, feat, labels, batch):

    # CE loss
    loss = torch.nn.functional.cross_entropy(logits, labels)

    # same original msg
    msg_ids = batch['msg_id'].to(device)

    same = (msg_ids.unsqueeze(0) == msg_ids.unsqueeze(1)).float()
    same.fill_diagonal_(0)

    # contrastive loss
    f = torch.nn.functional.normalize(feat, dim=1)
    sim = torch.matmul(f, f.T) / 0.1

    exp_sim = torch.exp(sim)

    pos = exp_sim * same
    neg = exp_sim * (1 - same)

    pos_sum = pos.sum(1)
    valid = (pos_sum > 0)

    if valid.sum() > 0:
        contrastive = -torch.log(
            pos_sum[valid] /
            (pos_sum[valid] + neg.sum(1)[valid] + 1e-8)
        ).mean()

        loss += 0.1 * contrastive

    return loss

# =========================================================
# 7. TRAIN
# =========================================================

def train_charcnn_m4(train_df):

    model = CharCNN().to(device)

    optimizer = torch.optim.Adam(
        model.parameters(),
        lr=1e-3
    )

    loader = DataLoader(
        SpamDataset(train_df),
        batch_size=64,
        shuffle=True
    )

    for epoch in range(3):
        model.train()

        for batch in loader:
            optimizer.zero_grad()

            logits, feat = model(
                batch['char_ids'].to(device)
            )

            loss = compute_m4_loss(
                logits,
                feat,
                batch['label'].to(device),
                batch
            )

            loss.backward()
            optimizer.step()

    return model

# =========================================================
# 8. EVALUATE
# =========================================================

def evaluate(model, df_eval):

    loader = DataLoader(
        SpamDataset(df_eval),
        batch_size=256
    )

    y_true, y_pred, types = [], [], []

    with torch.no_grad():
        for batch in loader:
            logits, _ = model(
                batch['char_ids'].to(device)
            )

            pred = torch.argmax(logits, 1).cpu().numpy()

            y_pred.extend(pred)
            y_true.extend(batch['label'].numpy())
            types.extend(batch['type'])

    y_true = np.array(y_true)
    y_pred = np.array(y_pred)
    types = np.array(types)

    def calc(mask):
        if mask.sum() == 0:
            return 0

        p, r, f1, _ = precision_recall_fscore_support(
            y_true[mask],
            y_pred[mask],
            average='binary',
            zero_division=0
        )
        return f1

    f1_all = calc(np.ones(len(y_true), dtype=bool))
    f1_real = calc(types == 'real')
    f1_syn = calc(types == 'synthetic')

    gap = abs(f1_real - f1_syn)

    return {
        "F1_all": f1_all,
        "F1_real": f1_real,
        "F1_syn": f1_syn,
        "Gap": gap
    }

# =========================================================
# 9. EXPERIMENT:
# TRY DIFFERENT SYNTHETIC SIZES
# =========================================================

synthetic_sizes = [ 0, 250,
    500,
    750,
    1000,
    1250,
    1500,
    1750,
    2000,
    2250,
    2500,
    2750,
    3000,
    3250,
    3500,
    3750,
    4000,4250,4500,4750,5000,5250,5500,5750,6000
    # 12000
]

results = []

for n_syn in synthetic_sizes:

    print(f"\n==============================")
    print(f"Running with {n_syn} synthetic samples")
    print(f"==============================")

    train_df = create_dataset_with_n_synthetic(
        train_ids,
        n_syn
    )

    model = train_charcnn_m4(train_df)

    res = evaluate(model, df_test)

    print(res)

    results.append({
        "Synthetic_Size": n_syn,
        **res
    })

# =========================================================
# 10. RESULT TABLE
# =========================================================

df_result = pd.DataFrame(results)

print("\n===== FINAL RESULT =====")
print(df_result)

# =========================================================
# 11. PLOT
# =========================================================

plt.figure(figsize=(10,6))

plt.plot(
    df_result["Synthetic_Size"],
    df_result["F1_real"],
    marker='o',
    label="F1_real"
)

plt.plot(
    df_result["Synthetic_Size"],
    df_result["F1_all"],
    marker='o',
    label="F1_all"
)

plt.plot(
    df_result["Synthetic_Size"],
    df_result["Gap"],
    marker='o',
    label="Gap (real-syn)"
)

plt.xlabel("Number of Synthetic Samples")
plt.ylabel("Score")
plt.title("CharCNN-M4: Optimal Synthetic Data Size")
plt.legend()
plt.grid()
plt.show()

# =========================================================
# Interpretation:
#
# Best synthetic size =
# highest F1_real
# + high F1_all
# + smallest Gap
#
# Usually NOT the largest synthetic size.
# Too much synthetic -> noise + semantic drift
# =========================================================
```

```python

# Run CharCNN-M4 10 times for each synthetic size and plot mean/std

synthetic_sizes = [0, 250,
    500,
    750,
    1000,
    1250,
    1500,
    1750,
    2000,
    2250,
    2500,
    2750,
    3000,
    3250,
    3500,
    3750,
    4000,4250,4500,4750,5000,5250,5500,5750,6000]
n_runs = 10

final_results = []

for n_syn in synthetic_sizes:
    print(f"\n===== Synthetic Size: {n_syn} =====")

    run_scores = []

    for run in range(n_runs):
        print(f"Run {run+1}/{n_runs}")

        # different seed each run
        torch.manual_seed(42 + run)
        np.random.seed(42 + run)

        train_df = create_dataset_with_n_synthetic(train_ids, n_syn)

        model = train_charcnn_m4(train_df)

        res = evaluate(model, df_test)

        run_scores.append(res["F1_real"])

        print(f"F1_real = {res['F1_real']:.4f}")

    mean_f1 = np.mean(run_scores)
    std_f1 = np.std(run_scores)

    final_results.append({
        "Synthetic_Size": n_syn,
        "Mean_F1_real": mean_f1,
        "Std_F1_real": std_f1
    })

df_final = pd.DataFrame(final_results)

print("\n===== FINAL SUMMARY =====")
print(df_final)

best_row = df_final.loc[df_final["Mean_F1_real"].idxmax()]
print("\nBest Synthetic Size =", int(best_row["Synthetic_Size"]))
print("Best Mean F1_real =", round(best_row["Mean_F1_real"], 4))

plt.figure(figsize=(10,6))

plt.errorbar(
    df_final["Synthetic_Size"],
    df_final["Mean_F1_real"],
    yerr=df_final["Std_F1_real"],
    marker="o",
    capsize=5
)

plt.xlabel("Synthetic Data Size")
plt.ylabel("Mean F1_real (10 runs)")
plt.title("CharCNN-M4: Optimal Synthetic Size (10 Runs Average)")
plt.grid(True)
plt.show()
```
