```python

# =========================================
# FINAL PIPELINE (GRID SEARCH + RESEARCH GRADE)
# =========================================

import torch, os, time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from tqdm import tqdm
from sklearn.metrics import precision_recall_fscore_support
from sklearn.model_selection import train_test_split, ParameterGrid
from transformers import AutoTokenizer, AutoModel, get_linear_schedule_with_warmup

# =========================================
# 1. LOAD DATA
# =========================================
df_stage2 = pd.read_csv('/kaggle/input/datasets/ngohoangdang/df-stage-2-out/df_stage2_out.csv')
df = pd.read_csv('/kaggle/input/datasets/organizations/uciml/sms-spam-collection-dataset/spam.csv', encoding='latin-1')

df_stage2 = df_stage2.drop_duplicates(subset=['obfuscated'])
df_stage2 = df_stage2[df_stage2['recover_score'] > 0.2]

df = df.reset_index().rename(columns={'index': 'msg_id'})
msg_map = df.set_index('v2')['msg_id'].to_dict()
df_stage2['msg_id'] = df_stage2['original'].map(msg_map)

train_ids, test_ids = train_test_split(df['msg_id'].unique(), test_size=0.2, random_state=42)
train_ids, val_ids = train_test_split(train_ids, test_size=0.1, random_state=42)

def create_dataset(ids):
    real = df[df['msg_id'].isin(ids)].copy()
    real['data_type'] = 'real'
    syn = df_stage2[df_stage2['msg_id'].isin(ids)].copy()
    syn['data_type'] = 'synthetic'

    full = pd.concat([
        real[['msg_id','v2','v1','data_type']].rename(columns={'v2':'text','v1':'label'}),
        syn[['msg_id','obfuscated','data_type']].rename(columns={'obfuscated':'text'})
    ])

    full.loc[full['data_type']=='synthetic','label'] = 'spam'
    return full

df_train = create_dataset(train_ids)
df_val   = create_dataset(val_ids)
df_test  = create_dataset(test_ids)

# =========================================
# 2. DEVICE
# =========================================
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
NUM_GPU = torch.cuda.device_count()

SAVE_DIR = "./models"
os.makedirs(SAVE_DIR, exist_ok=True)

# =========================================
# 3. TOKENIZER + CHAR
# =========================================
tokenizer = AutoTokenizer.from_pretrained("distilbert-base-uncased")

ALL_CHARS = list("abcdefghijklmnopqrstuvwxyz0123456789!@#$%^&*()-_=+[]{};:'\",.<>/? ")
char2idx = {c:i+1 for i,c in enumerate(ALL_CHARS)}

def encode_char(text, max_len=128):
    arr = [char2idx.get(c.lower(),0) for c in text[:max_len]]
    arr += [0]*(max_len-len(arr))
    return arr

# =========================================
# 4. DATASET
# =========================================
class SpamDataset(torch.utils.data.Dataset):
    def __init__(self, df):
        self.df = df.reset_index(drop=True)
        self.map = {'ham':0,'spam':1}

    def __len__(self): return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]

        enc = tokenizer(str(row['text']),
                        truncation=True,
                        padding='max_length',
                        max_length=64,
                        return_tensors='pt')

        return {
            'ids': enc['input_ids'].squeeze(),
            'mask': enc['attention_mask'].squeeze(),
            'char_ids': torch.tensor(encode_char(str(row['text']))),
            'label': torch.tensor(self.map[row['label']]),
            'type': row['data_type']
        }

# =========================================
# 5. MODELS
# =========================================
class DistilBERT_Model(torch.nn.Module):
    def __init__(self):
        super().__init__()
        self.bert = AutoModel.from_pretrained("distilbert-base-uncased")
        self.fc = torch.nn.Linear(768,2)
    def forward(self, ids, mask, char_ids=None):
        out = self.bert(ids, attention_mask=mask)
        feat = out.last_hidden_state[:,0,:]
        return self.fc(feat), feat

class FastTextModel(torch.nn.Module):
    def __init__(self):
        super().__init__()
        self.emb = torch.nn.Embedding(30522,128)
        self.fc = torch.nn.Linear(128,2)
    def forward(self, ids, mask=None, char_ids=None):
        x = self.emb(ids)
        feat = x.mean(1)
        return self.fc(feat), feat

class TextCNN(torch.nn.Module):
    def __init__(self):
        super().__init__()
        self.emb = torch.nn.Embedding(30522,128)
        self.convs = torch.nn.ModuleList([torch.nn.Conv1d(128,128,k) for k in [3,4,5]])
        self.fc = torch.nn.Linear(384,2)
    def forward(self, ids, mask=None, char_ids=None):
        x = self.emb(ids).permute(0,2,1)
        xs = [torch.relu(c(x)).max(2)[0] for c in self.convs]
        feat = torch.cat(xs,1)
        return self.fc(feat), feat

class CharCNN(torch.nn.Module):
    def __init__(self):
        super().__init__()
        self.emb = torch.nn.Embedding(len(char2idx)+1,64)
        self.conv = torch.nn.Sequential(
            torch.nn.Conv1d(64,128,3),
            torch.nn.ReLU(),
            torch.nn.Conv1d(128,128,3),
            torch.nn.ReLU()
        )
        self.fc = torch.nn.Linear(128,2)
    def forward(self, ids=None, mask=None, char_ids=None):
        x = self.emb(char_ids).permute(0,2,1)
        x = self.conv(x)
        feat = x.max(2)[0]
        return self.fc(feat), feat

# =========================================
# 6. LOSS (WEIGHTED)
# =========================================
def compute_loss(logits, feat, labels, m, w):
    loss = torch.nn.functional.cross_entropy(logits, labels)

    if m >= 2:
        loss += w['alpha'] * torch.nn.functional.kl_div(
            torch.log_softmax(logits,1),
            torch.softmax(logits.detach(),1),
            reduction='batchmean'
        )

    if m >= 3:
        p = torch.softmax(logits,1)
        loss += w['eta'] * -(p * torch.log(p+1e-8)).sum(1).mean()

    if m == 4:
        loss += w['beta'] * torch.nn.functional.mse_loss(feat, feat.mean(0))
        loss += w['gamma'] * torch.var(feat, dim=0).mean()

    return loss

# =========================================
# 7. TRAIN
# =========================================
def train_model(model, train_df, val_df, m, w, patience=2):

    if NUM_GPU > 1:
        model = torch.nn.DataParallel(model)

    model.to(device)

    train_loader = torch.utils.data.DataLoader(SpamDataset(train_df), batch_size=64, shuffle=True)
    val_loader   = torch.utils.data.DataLoader(SpamDataset(val_df), batch_size=128)

    opt = torch.optim.AdamW(model.parameters(), lr=3e-5)

    total_steps = len(train_loader)*10
    scheduler = get_linear_schedule_with_warmup(opt, int(0.1*total_steps), total_steps)

    scaler = torch.cuda.amp.GradScaler()

    best_f1, patience_counter = -1, 0
    best_state = None

    loss_curve = []

    for epoch in range(10):
        model.train()
        total = 0

        for b in train_loader:
            opt.zero_grad()

            with torch.cuda.amp.autocast():
                logits, feat = model(
                    b['ids'].to(device),
                    b['mask'].to(device),
                    b['char_ids'].to(device)
                )
                loss = compute_loss(logits, feat, b['label'].to(device), m, w)

            scaler.scale(loss).backward()
            scaler.step(opt)
            scaler.update()
            scheduler.step()

            total += loss.item()

        loss_curve.append(total/len(train_loader))

        # VALIDATION
        model.eval()
        y_true, y_pred = [],[]

        with torch.no_grad():
            for b in val_loader:
                logits,_ = model(
                    b['ids'].to(device),
                    b['mask'].to(device),
                    b['char_ids'].to(device)
                )
                pred = torch.argmax(logits,1).cpu().numpy()
                y_pred.extend(pred)
                y_true.extend(b['label'].numpy())

        f1 = precision_recall_fscore_support(y_true,y_pred,average='binary',zero_division=0)[2]

        if f1 > best_f1:
            best_f1 = f1
            best_state = model.state_dict()
            patience_counter = 0
        else:
            patience_counter += 1

        if patience_counter >= patience:
            break

    model.load_state_dict(best_state)
    return model, loss_curve, best_f1

# =========================================
# 8. EVALUATE
# =========================================
def evaluate(model, df):
    loader = torch.utils.data.DataLoader(SpamDataset(df), batch_size=128)

    y_true,y_pred,types = [],[],[]
    start = time.time()

    with torch.no_grad():
        for b in loader:
            logits,_ = model(
                b['ids'].to(device),
                b['mask'].to(device),
                b['char_ids'].to(device)
            )

            pred = torch.argmax(logits,1).cpu().numpy()
            y_pred.extend(pred)
            y_true.extend(b['label'].numpy())
            types.extend(b['type'])

    latency = time.time() - start

    def calc(mask):
        yt = np.array(y_true)[mask]
        yp = np.array(y_pred)[mask]
        return precision_recall_fscore_support(yt,yp,average='binary',zero_division=0)[:3]

    return {
        "all": calc(np.ones(len(y_true),bool)),
        "real": calc(np.array(types)=='real'),
        "syn": calc(np.array(types)=='synthetic'),
        "latency": latency
    }

# =========================================
# 9. GRID SEARCH SPACE
# =========================================
grid = {
    'alpha':[0.5,1.0],
    'beta':[0.1,0.3],
    'gamma':[0.1],
    'eta':[0.1]
}

# =========================================
# 10. TRAIN ALL MODELS
# =========================================
models = {
    "DistilBERT":DistilBERT_Model,
    "FastText":FastTextModel,
    "TextCNN":TextCNN,
    "CharCNN":CharCNN
}

results = {}
loss_curves = {}

for name, cls in models.items():
    for m in [1,2,3,4]:

        tag = f"{name}_M{m}"
        print("\n===", tag, "===")

        train_df_use = df_train if m!=1 else df_train[df_train.data_type=='real']

        if m == 4:
            best_score = -1

            for w in ParameterGrid(grid):
                model = cls()
                model,curve,score = train_model(model, train_df_use, df_val, m, w)

                if score > best_score:
                    best_model = model
                    best_curve = curve
                    best_score = score
                    best_w = w

            print("Best weights:", best_w)
            model = best_model
            curve = best_curve
        else:
            w = {'alpha':1,'beta':0.1,'gamma':0.1,'eta':0.1}
            model = cls()
            model,curve,_ = train_model(model, train_df_use, df_val, m, w)

        torch.save(model.state_dict(), f"{SAVE_DIR}/{tag}.pt")

        results[tag] = evaluate(model, df_test)
        loss_curves[tag] = curve

# =========================================
# 11. RESULT TABLE
# =========================================
rows=[]
for k,v in results.items():
    rows.append({
        "Model":k,
        "F1_all":v['all'][2],
        "F1_real":v['real'][2],
        "F1_syn":v['syn'][2],
        "Latency":v['latency']
    })

df_res = pd.DataFrame(rows)
print(df_res)

# =========================================
# 12. LOSS CURVES
# =========================================
plt.figure(figsize=(12,6))
for k,v in loss_curves.items():
    plt.plot(v,label=k)
plt.legend()
plt.title("Loss Curves")
plt.show()
```
